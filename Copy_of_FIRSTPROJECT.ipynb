```python
from crewai import Agent, Task, Crew, Process
researcher= Agent (
    role= " Senior Researcher Analyst     ",
    goal= "   Discover groundbreaking technologies in AI and data Scince     ",
    backstory= " An expert analyst with a knack for finding hidden gems in data.        "
                   )
writer= Agent(
    role= " Tech Content Strategist         ",
    goal= " Craft complling content on technical advancement       ",
    backstory= " A skilled writer who can distill complex topics into clear narratives.         "
)
research_task=Task(
  description= " Identify key trends in the use of AI in healthcare for 2026.  ",
  agent=researcher,
  expected_output= "A comprehensive report on key trends in AI in healthcare for 2026 including key trends, emerging technologies, and their potential impact."
)
writer_task=Task(
  description= "Compose a 3-paragraph blog post summarizing the key trends found.    ",
  agent=writer,
  expected_output= "A 3-paragraph blog post, written in an engaging and informative style, summarizing the key trends in AI in healthcare for 2026, suitable for a tech-focused audience."
)
```

```python
research_crew= Crew(
    agents=[researcher, writer],
    tasks=[research_task, writer_task],
    process= Process.sequential
)
```

```python
import asyncio


async def run_crew():
    result= await research_crew.kickoff_async()
    print("############")
    print(result)

# Run the asynchronous function
await run_crew()
```
